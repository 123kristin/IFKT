```python
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
path = "algebra_2006_2007_train.txt"
df = pd.read_table(path, encoding="ISO-8859-15", low_memory=False)
df.shape
```

```python
df.columns
```

```python

df.dropna(subset=['Problem Name', 'KC(Default)', 'Step Duration (sec)', 'Step Duration (sec)'], inplace=True)
df
```

```python
key_user, key_qtype, key_q, key_s, key_elpsed, key_correct = 'Anon Student Id', 'Problem Hierarchy', 'Problem Name', 'KC(Default)', 'Step Duration (sec)', 'Correct First Attempt'
columns = [key_user, key_qtype, key_q, key_s, key_elpsed, key_correct]
df1 = df[[key_user, key_qtype, key_q, key_s, key_elpsed, key_correct]]
df1.head()
```

### algbra interaction

```python
len(df1)
```

```python
question_id_dict = dict(zip(df[key_q].unique(), range(len(df[key_q].unique()))))
question_type_dict = dict(zip(df[key_qtype].unique(), range(len(df[key_qtype].unique()))))
user_id_dict = dict(zip(df[key_user].unique(), range(len(df[key_user].unique()))))
skill_id_dict = dict(zip(df[key_s].unique(), range(len(df[key_s].unique()))))

df1[key_q] = df1[key_q].map(question_id_dict)
df1[key_s] = df1[key_s].map(skill_id_dict)
df1[key_qtype] = df1[key_qtype].map(question_type_dict)

N_QUESTION = len(question_id_dict)
N_SKILL = len(skill_id_dict)
N_QUESTION_TYPE = len(question_type_dict)
N_USER = len(user_id_dict)

print("N_USER:{}, N_QUESTION:{}, N_SKILL:{}, N_QUESTION_TYPE:{}".format(N_USER, N_QUESTION, N_SKILL, N_QUESTION_TYPE))
```

```python
key_qd = 'diff'

group1 = df1[[key_q, key_correct]].groupby([key_q]).apply(lambda r:r[key_correct].sum() / len(r[key_correct]))
group1
df1[key_qd] = df1[key_q].map(group1)
df1.head()
```

```python
# 我们需要将数据进行预处理，每个学生的学习记录利用group by合并为序列。
group = df1.groupby([key_user]).apply(lambda r: (
            r[key_q].values,
            r[key_s].values,
            r[key_qtype].values,
            r[key_qd].values,
            r[key_elpsed].values,
            r[key_correct].values
            ))
```

```python
print(len(group))
```

```python
df1.to_csv("data.csv", index=None)
```

```python
from tqdm import tqdm
from sklearn.model_selection import train_test_split
```

```python
from tqdm import tqdm
from sklearn.model_selection import train_test_split

train, test = train_test_split(group, test_size=0.2, random_state=42)

def generate_df_by_group(group):
    df = pd.DataFrame(columns=['user_id', 'q_idx', 's_idx', 'q_type', 'q_diff', 'ms_first_response', 'correct'])
    for user_id, (q, s, qt, qd, ms_first_response, correct) in tqdm(group.items()):
        for i in range(len(q)):
            tmp = [user_id, q[i], s[i], qt[i], qd[i], ms_first_response[i], correct[i]]
            df.loc[len(df)] = tmp
    return df
print(len(train), len(test))
```

```python
df_train = generate_df_by_group(train)
df_test = generate_df_by_group(test)
```

```python
from sklearn.preprocessing import MinMaxScaler

sacler = MinMaxScaler()

col = ['ms_first_response']
tmp = sacler.fit_transform(df_train[col])
df_train[col] = tmp

tmp = sacler.transform(df_test[col])
df_test[col] = tmp
```
