```python
data = pd.read_csv('processed_data.csv', encoding="ISO-8859-1", low_memory=False)
data.head(), data.shape, data.columns
```

```python
key_s = 'skill_id'
skill_id_dict = dict(zip(data['KC(Default)'].unique(), range(len(data['KC(Default)'].unique()))))
data[key_s] = data['KC(Default)'].map(skill_id_dict)
data.head(), data.shape, data.columns
```

```python
df = data.rename({'diff':'difficulty'}, axis=1)
df.shape, df.columns
```
